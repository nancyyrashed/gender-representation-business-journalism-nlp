### **2.1 Defining scraping and extraction functions**

A CSV file contains the list of URLs that need to be examined for this project. We must build a function that we can execute to scrape the data as we iterate through the URLs before importing this file and extracting the content from the URLs. Since the HTML on each page we plan to scrape differs, we must also create a unique function for each to retrieve the necessary data.

The below function checks if a webpage is accessible and returns the content of it:


```python
def getWebpage(url):
    
    #checks if the page can be scraped
    
    request_headers = {}
    
    response = requests.get(url, headers=request_headers)
    
    if response.status_code != 200:
        soup = 'error'
    else:
        content = response.content
        
        #get the webpage now as it can be scraped
        soup = BeautifulSoup(response.content, 'html.parser')
        
    return soup
```

### **Extracting information from The Guardian webpages**

The following function extracts title, author and article text from a Guardian webpage:


```python
def getDataFromGurdian(soup):
    
    #title of the article 
    title = soup.find(['h1'])
    
    #author of the article
    author = soup.find('a', {'rel': 'author'})
    
    #article text
    article = soup.find('div', {'id': 'maincontent'}).find_all("p")
    
    #produce the article as one string without spaces
    article_text = ''
    for info in article:
        article_text = article_text + ' ' + info.get_text().lstrip().rstrip()
        
    return title.get_text(), author.get_text(), article_text[1:]
```

### **Extracting information from The Prestige webpages**

The following function extracts title, author and article text from a prestige webpage:

```python
def getDataFromChannelNewsAsia(soup):
    
    #title of the article 
    title = soup.find(['h1'])
    
    #author of the article
    author = soup.find('div', {'class': 'mobile_author_card__name'}).find('a')
    
    #article text
    article = soup.find('section', {'class': 'block block-layout-builder block-field-blocknodearticlefield-content clearfix'}).find_all("p")
    
    #produce the article as one string without spaces
    article_text = ''
    for info in article:
        article_text = article_text + ' ' + info.get_text().lstrip().rstrip()
        
    return title.get_text(), author.get_text(), article_text[1:]
```

### **Extracting information from The Forbes webpages**

The following function extracts title, author and article text from a forbes webpage:

```python
def getDataFromForbes(soup):
    
    #title of the article 
    title = soup.find(['h1'])
    
    #author of the article
    author = soup.find('span', {'class': 'fs-author-name'}).find('a')
    
    #article text
    article = soup.find('div', {'class': 'article-body fs-article fs-responsive-text current-article'}).find_all("p")
    
    #produce the article as one string without spaces
    article_text = ''
    for info in article:
        article_text = article_text + ' ' + info.get_text().lstrip().rstrip()
        
    return title.get_text(), author.get_text(), article_text[1:]
```

### **Extracting information from The britannica webpages**

The following function extracts title, author and article text from a britannica webpage:

```python
def getDataFromBritannica(soup):
    
    #title of the article 
    title = soup.find(['h1'])
    
    #author of the article
    author = soup.find('span',class_="btn btn-link byline-link editor-link p-0 gtm-byline font-12 byline-contributor" )
    
    #article text
    article = soup.find('section', {'id': 'ref1', 'data-level': '1'}).find_all("p")
    
    #produce the article as one string without spaces
    article_text = ''
    for info in article:
        article_text = article_text + ' ' + info.get_text().lstrip().rstrip()
        
    return title.get_text(), author.get_text(), article_text[1:]
```

### **2.2.2 Scrape data and save to dataframe**

Now, using the functions previously defined, we will try to scrape the articles at each of the URLs we have imported. If it worked, we will attach the information that was scraped to a dataframe.

### **3.2.1 Removing stop words using nltk and calculating the frequency of words**

there are some words that aren't important in the analysis that should be removed to be able to proceed. i will use nltk stop words to be able to remove them and then calculate the frequency of words after:

```python
def get_freqs(words):
    # Remove certain punctuation using regex. This will stop 'said' and 'said.' appearing as different words
    # Do not remove apostrophes
    words = re.sub('\.|\,|\"|\“|\”|\—|\–|\-|\?|\!|\:|\;|\(|\)', '', words)
    
    # Split the corpus into a list of words
    words_list = words.lower().split()
  
    # Converting to a set means all duplicate words are removed
    unique_words_full = set(words_list)
    
    # Remove stop words
    unique_words = [word for word in unique_words_full if word.lower() not in stopwords.words('english')]
    
    # Use Counter to count word frequencies
    word_freqs = Counter(words_list)
    
    # Filter out stop words
    word_freqs = {word: freq for word, freq in word_freqs.items() if word.lower() not in stopwords.words('english')}
    
    return word_freqs

first_freq = getFreqs(editors['Full words'][0])
print('1st:', first_freq)

second_freq = getFreqs(editors['Full words'][1])
print('2nd:', second_freq)

third_freq = getFreqs(editors['Full words'][2])
print('3rd:', third_freq)

fourth_freq = getFreqs(editors['Full words'][3])
print('4th:', fourth_freq)
```

### **3.2.2 Filtered words**

now we will filter it even more by removing punctuation, or suffixes separated from words:

```python
def removeStopwords(words):
    words = words.lower()
    
    # Remove certain punctuation using regex. This will stop 'said' and 'said.' appearing as different words
    # Do not remove apostrophes
    words = re.sub('\.|\,|\"|\“|\”|\—|\–|\-|\?|\!|\:|\;|\/|\(|\)', '', words)
    
    # Tokenize the corpus - this creates a list of all the words in the corpus 
    tokens = word_tokenize(words)
    
    # As well as stopwords, we want to remove tokens with only punctuation, or suffixes separated from words
    punct = ["'", "''", '``', '(', ')', '%', '&', '...', '…', "‘", "’"]
    suffixes = ["'s", "n't", "'ve", "'ll", "'re", "'d"]
    remove_words = punct + suffixes + stopwords.words('english')
    
    # Create a new list with stop words removed from the corpus
    filtered_tokens = [word for word in tokens if not word in remove_words]
 
    # Reconstruct corpus as a string
    filtered_words = ' '.join(filtered_tokens)
    
    return filtered_words

# Filter the editors corpora and add as a new column
editors['Filtered words'] = ''
editors['Filtered title words'] = ''

for i in range(0, len(editors)):
    editors.loc[i, 'Filtered words'] = removeStopwords(editors['Full words'][i])
    editors.loc[i, 'Filtered title words'] = removeStopwords(editors['Title words'][i])

# Filter the articles corpora and add as a new column
scrapData['Filtered text'] = ''
scrapData['Filtered title'] = ''

for i in range(0, len(scrapData)):
    scrapData.loc[i, 'Filtered text'] = removeStopwords(scrapData['Text'][i])
    scrapData.loc[i, 'Filtered title'] = removeStopwords(scrapData['Title'][i])

# Display the filtered words for the four articles
for i in range(len(scrapData)):
    # Represents the filtered words in the article text
    print(f"\nFiltered words for Article {i + 1}:")
    print(scrapData.loc[i, 'Filtered text'])
    print({len(scrapData.loc[i, 'Filtered text'])})
    # Represents the filtered words in the article title
    print("\nFiltered title words for Article:")
    print(scrapData.loc[i, 'Filtered title'])
    print({len(scrapData.loc[i, 'Filtered title'])})
    print("\n" + "="*50 + "\n")
```

### **3.3 Lemmatization**

In NLTK, lemmatization refers to the computational procedure that determines a word's lemma based on its context and meaning. Typically, lemmatization refers to the structural analysis of words with the intention of eliminating inflectional ends. It aids in providing the lemma, or base or dictionary form, of a word. For example, reducing "builds", "building", or "built" to the lemma "build". so it is better to lemmatize the words to be able to proceed with the analysis:

```python
lemmatizer = WordNetLemmatizer()

def lemmatizeWords(words):
    tokens = word_tokenize(words)
    
    lemmatized_tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    return ' '.join(lemmatized_tokens)

# Lemmatize the editors words and add as a new column
editors['Lemmatized words']=''
editors['Lemmatized title words']=''

for i in range(0, len(editors)):
    editors.loc[i, 'Lemmatized words'] = lemmatizeWords(editors['Filtered words'][i])
    editors.loc[i, 'Lemmatized title words'] = lemmatizeWords(editors['Filtered title words'][i])
    
# Print lemmatized words and title for all editors
for i in range(len(editors)):
    print(f"\nLemmatized text for Article {i + 1}:")
    print(editors.loc[i, 'Lemmatized words'])
    print(f"Length: {len(editors.loc[i, 'Lemmatized words'])}")

    # Represents the lemmatized title in the article
    print("\nLemmatized title for Article:")
    print(editors.loc[i, 'Lemmatized title words'])
    print(f"Length: {len(editors.loc[i, 'Lemmatized title words'])}")

    print("\n" + "="*50 + "\n")
```

## **4 Generating features**

### **4.1 Average sentence length**

this is a feature that calculates each article's average sentence length. Sentence length might reveal information about the text's complexity and writing style. This code first tokenizes the text into sentences and words using NLTK, and then it determines the average sentence length in words: 

```python
# Function to calculate average sentence length
def calculateAvgSentenceLength(text):
    sentences = nltk.sent_tokenize(text)
    total_words = len(word_tokenize(text))
    total_sentences = len(sentences)
    
    if total_sentences > 0:
        avg_sentence_length = total_words / total_sentences
        return avg_sentence_length
    else:
        return 0

# Create a new column for average sentence length
scrapData['Avg Sentence Length'] = scrapData['Text'].apply(calculateAvgSentenceLength)

scrapData['Avg Sentence Length']
```

### **4.2 Ratio of unique words**

Using a text input, this feature determines the proportion of unique words to total words. After tokenizing the text into individual words, it counts the total number of words, uses a set to identify unique words, and computes the unique word ratio:

```python
# Function to calculate the ratio of unique words to total words
def calculateUniqueWordRatio(text):
    words = word_tokenize(text)
    unique_words = set(words)
    total_words = len(words)
    
    if total_words > 0:
        unique_word_ratio = len(unique_words) / total_words
        return unique_word_ratio
    else:
        return 0

# Create a new column for the unique word ratio
scrapData['Unique Word Ratio'] = scrapData['Text'].apply(calculateUniqueWordRatio)

scrapData['Unique Word Ratio']
```

### **4.3 Reading Ease**

This feature takes a text input and calculates the Flesch Reading Ease score using the Textatistic library, The Flesch Reading Ease is a numerical score that quantifies the readability of a piece of text:

```python
# Function to calculate Flesch Reading Ease score
def calculateFleschReadingEase(text):
    try:
        reading_ease_score = Textatistic(text).scores['flesch_score']
        return reading_ease_score
    except Exception as e:
        print(f"Error calculating Flesch Reading Ease: {e}")
        return 0

# Create a new column for Flesch Reading Ease score
scrapData['Flesch Reading Ease'] = scrapData['Text'].apply(calculateFleschReadingEase)

# Display the Flesch Reading Ease scores
print(scrapData['Flesch Reading Ease'])
```

### **4.3 Sentiment analysis**

This function takes a text input and calculates the ratio of positive to negative sentiment words using the opinion lexicon dataset from NLTK, it is for assessing the overall sentiment bias in a piece of text based on the occurrences of positive and negative words, If the sentiment bias ratio is greater than 1 it indicates a higher proportion of positive sentiment words compared to negative sentiment words:

```python
# List of positive and negative words from the opinion lexicon
positive_words = set(opinion_lexicon.positive())
negative_words = set(opinion_lexicon.negative())

# Function to calculate the ratio of positive to negative sentiment words
def calculateSentimentBiasRatio(text):
    words = word_tokenize(text.lower())
    positive_count = sum(1 for word in words if word in positive_words)
    negative_count = sum(1 for word in words if word in negative_words)
    
    if negative_count > 0:
        sentiment_ratio = positive_count / negative_count
        return sentiment_ratio
    else:
        return 0

# Create a new column for sentiment bias ratio
scrapData['Sentiment Bias Ratio'] = scrapData['Text'].apply(calculateSentimentBiasRatio)

scrapData['Sentiment Bias Ratio']
```

### **4.4 Quotations from gender**

This function takes a text input and calculates the ratio of quotes from male sources to the total number of quotes (including both male and female sources), If the ratio is less than 0.5, it indicates a higher proportion of quotes from female sources compared to male sources:

```python
# Function to calculate the ratio of quotes from male and female sources
def calculateGenderQuotedSourcesRatio(text):
    male_quotes = re.findall(r'\bhe\b|\bhis\b|\bhim\b', text.lower())
    female_quotes = re.findall(r'\bshe\b|\bher\b', text.lower())
    
    total_quotes = len(male_quotes) + len(female_quotes)
    
    if total_quotes > 0:
        gender_ratio = len(male_quotes) / total_quotes
        return gender_ratio
    else:
        return 0

# Create a new column for gender quoted sources ratio
scrapData['Gender Quoted Sources Ratio'] = scrapData['Text'].apply(calculateGenderQuotedSourcesRatio)

scrapData['Gender Quoted Sources Ratio']
```

### **4.5 Subjectivity Score**

The Subjectivity Score calculates how subjective or opinionated a given text is. It offers a subjectivity score of 0 to 1, with 0 indicating highly objective (factual) content and 1 indicating very subjective (opinionated) content. It is computed using TextBlob:

```python
# Function to calculate subjectivity score
def calculateSubjectivity(text):
    try:
        blob = TextBlob(text)
        subjectivity_score = blob.sentiment.subjectivity
        return subjectivity_score
    except Exception as e:
        print(f"Error calculating Subjectivity Score: {e}")
        return 0

# Create a new column for Subjectivity Score
scrapData['Subjectivity Score'] = scrapData['Text'].apply(calculateSubjectivity)

scrapData['Subjectivity Score']
```

### **5.4 Scatter Plot: Text Length vs. Average Sentence Length**

This scatter plot shows the relationship between the two variables and provides information about the ranges in average sentence length with text length:

```python
# Function to calculate average sentence length
def calculateAverageSentenceLength(text):
    try:
        sentences = nltk.sent_tokenize(text)
        average_sentence_length = sum(len(nltk.word_tokenize(sentence)) for sentence in sentences) / len(sentences) if len(sentences) > 0 else 0
        return average_sentence_length
    except Exception as e:
        print(f"Error calculating Average Sentence Length: {e}")
        return 0

# Calculate text length and average sentence length for each article
scrapData['Text Length'] = scrapData['Text'].apply(lambda x: len(x.split()))
scrapData['Avg Sentence Length'] = scrapData['Text'].apply(calculateAverageSentenceLength)

# Plotting the scatter plot
plt.figure(figsize=(14, 6))
plt.scatter(scrapData['Text Length'], scrapData['Avg Sentence Length'], color='skyblue')
plt.xlabel('Text Length (Number of Words)')
plt.ylabel('Average Sentence Length (Number of Words)')
plt.title('Scatter Plot: Text Length vs. Average Sentence Length')
plt.show()
```
